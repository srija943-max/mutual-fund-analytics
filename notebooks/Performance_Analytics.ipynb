# Fund Performance Analytics Report

## Summary Metrics & Formulae Applied:
1. **Daily Returns**: Calculated using `(NAV_t / NAV_t-1) - 1`.
2. **CAGR**: Calculated annualized growth for 1yr, 3yr, and 5yr.
3. **Sharpe & Sortino Ratios**: Risk-adjusted returns calculated using Rf = 6.5%.
4. **Alpha & Beta**: Evaluated via OLS regression against Nifty 100 benchmark returns.
5. **Composite Fund Scorecard**: Created overall 0-100 composite rankings.

All outputs generated in `reports/fund_scorecard.csv`, `reports/alpha_beta.csv`, and `reports/benchmark_comparison_chart.png`.
