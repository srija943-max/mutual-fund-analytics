# Exploratory Data Analysis (EDA) Findings

1. **NAV Trend**: Highlighted 2023 bull run and 2024 market corrections across 40 schemes.
2. **AUM Growth**: Total AUM crossed ₹12.5L Cr with SBI dominating the market share.
3. **SIP Inflow**: Monthly SIP reached an all-time high of ₹31,002 Cr by Dec 2025.
4. **Category Inflows**: Equity and Index funds showed maximum color intensity in net monthly inflows.
5. **Investor Demographics**: 40% of investors fall in the 26-40 age group.
6. **Geographic Distribution**: T30 cities contribute 75% of investment volume compared to B30.
7. **Folio Growth**: Folios grew significantly from 13.26 Cr (Jan 2022) to 26.12 Cr (Dec 2025).
8. **NAV Correlation**: Strong positive return correlation observed among large-cap funds.
9. **Sector Allocation**: Banking and IT represent over 55% of portfolio holdings.
10. **Expense Ratio vs Return**: Lower expense ratios show consistent long-term alpha.
